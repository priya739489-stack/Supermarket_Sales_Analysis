```python
import pandas as pd

# Load your uploaded CSV file
df = pd.read_csv('Anyone with the link - Sheet1.csv')

# Display the first 5 rows to confirm it worked
display(df.head())
```

```python
# Check data types and non-null counts
df.info()
```

```python
# Display basic information about the columns and data types
df.info()

# Display summary statistics for numerical columns
df.describe()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Convert Order_Date to datetime format so Python understands dates properly
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%Y', errors='coerce')

# Sort the dataframe by date
df = df.sort_values('Order_Date')

# Group by date and sum up the revenue
daily_revenue = df.groupby('Order_Date')['Revenue'].sum()

# Plot the revenue trend
plt.figure(figsize=(10, 5))
plt.plot(daily_revenue.index, daily_revenue.values, marker='o', linestyle='-', color='b')
plt.title('E-Commerce Revenue Trend Over Time')
plt.xlabel('Order Date')
plt.ylabel('Total Revenue')
plt.grid(True)
plt.show()
```

```python
# Convert Revenue and Profit columns to clean numbers
df['Revenue'] = pd.to_numeric(df['Revenue'].astype(str).str.replace(',', '').str.replace('$', ''), errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'].astype(str).str.replace(',', '').str.replace('$', ''), errors='coerce')

# Group by Category and calculate total Revenue and Profit
category_performance = df.groupby('Category')[['Revenue', 'Profit']].sum()

# Display the performance table
display(category_performance)

# Plot a bar chart for category revenue
category_performance['Revenue'].plot(kind='bar', color='orange', figsize=(8, 4))
plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue')
plt.grid(axis='y')
plt.show()
```

```python
# Group by Customer_ID to calculate key metrics for churn analysis
customer_summary = df.groupby('Customer_ID').agg(
    Total_Spent=('Revenue', 'sum'),
    Order_Count=('Order_ID', 'count'),
    Last_Order_Date=('Order_Date', 'max')
).reset_index()

# Display the first 5 rows of our customer feature table
display(customer_summary.head())
```

```python
import datetime as dt

# Set a reference snapshot date (using the max date in your dataset as today)
snapshot_date = df['Order_Date'].max()

# Calculate days since last order for each customer
customer_summary['Recency'] = (snapshot_date - customer_summary['Last_Order_Date']).dt.days

# Define Churn: Let's flag customers as churned (1) if they haven't ordered in a while (e.g., > 30 days)
# Otherwise, active (0)
customer_summary['Churn'] = customer_summary['Recency'].apply(lambda x: 1 if x > 30 else 0)

# Display the final churn dataset head
display(customer_summary.head())

# Check the distribution of churned vs active customers
print(customer_summary['Churn'].value_counts())
```

```python
# Calculate Supermarket Sales Project Insights

print("=== SUPERMARKET SALES ANALYSIS REPORT ===\n")

# 1. Top Product by Sales
if 'Product' in df.columns and 'Sales' in df.columns:
    top_product = df.groupby('Product')['Sales'].sum().idxmax()
    top_product_sales = df.groupby('Product')['Sales'].sum().max()
    print(f"• Top Product: {top_product} (Sales: ₹{top_product_sales:,.2f})")

# 2. Best Branch / City Performance
if 'Branch' in df.columns and 'Sales' in df.columns:
    top_branch = df.groupby('Branch')['Sales'].sum().idxmax()
    top_branch_sales = df.groupby('Branch')['Sales'].sum().max()
    print(f"• Best Branch: {top_branch} (Sales: ₹{top_branch_sales:,.2f})")

# 3. Top Category
if 'Category' in df.columns and 'Sales' in df.columns:
    top_category = df.groupby('Category')['Sales'].sum().idxmax()
    top_category_sales = df.groupby('Category')['Sales'].sum().max()
    print(f"• Top Category: {top_category} (Sales: ₹{top_category_sales:,.2f})")

# 4. Most Popular Payment Method
if 'Payment' in df.columns or 'Payment_Method' in df.columns:
    pay_col = 'Payment' if 'Payment' in df.columns else 'Payment_Method'
    top_payment = df[pay_col].mode()[0]
    pay_count = df[pay_col].value_counts().max()
    print(f"• Most Popular Payment: {top_payment} ({pay_count} transactions)")

# 5. Customer Type Spending Comparison
if 'Customer_type' in df.columns and 'Sales' in df.columns:
    cust_spending = df.groupby('Customer_type')['Sales'].mean()
    print(f"• Customer Type Average Spending:\n{cust_spending.apply(lambda x: f'₹{x:,.2f}')}")

# 6. Average Customer Rating
if 'Rating' in df.columns:
    avg_rating = df['Rating'].mean()
    print(f"\n• Average Customer Rating: {avg_rating:.2f} out of 5")
```

```python
print("=== SUPERMARKET SALES ANALYSIS REPORT ===\n")

# Find the correct sales/revenue column name
sales_col = 'Revenue' if 'Revenue' in df.columns else ('Sales' if 'Sales' in df.columns else df.columns[0])

# 1. Top Product
if 'Product' in df.columns:
    top_prod = df.groupby('Product')[sales_col].sum().idxmax()
    top_prod_val = df.groupby('Product')[sales_col].sum().max()
    print(f"• Top Product: {top_prod} (Sales: ₹{top_prod_val:,.2f})")

# 2. Best Branch/Region
branch_col = 'Branch' if 'Branch' in df.columns else ('Region' if 'Region' in df.columns else None)
if branch_col:
    top_branch = df.groupby(branch_col)[sales_col].sum().idxmax()
    top_branch_val = df.groupby(branch_col)[sales_col].sum().max()
    print(f"• Best Branch/Region: {top_branch} (Sales: ₹{top_branch_val:,.2f})")

# 3. Top Category
if 'Category' in df.columns:
    top_cat = df.groupby('Category')[sales_col].sum().idxmax()
    top_cat_val = df.groupby('Category')[sales_col].sum().max()
    print(f"• Top Category: {top_cat} (Sales: ₹{top_cat_val:,.2f})")

# 4. Most Popular Payment Method
pay_col = next((col for col in ['Payment', 'Payment_Method', 'Payment Method'] if col in df.columns), None)
if pay_col:
    top_pay = df[pay_col].mode()[0]
    pay_count = df[pay_col].value_counts().max()
    print(f"• Most Popular Payment: {top_pay} ({pay_count} transactions)")

# 5. Average Customer Rating
if 'Rating' in df.columns:
    avg_rating = df['Rating'].mean()
    print(f"• Average Customer Rating: {avg_rating:.2f} out of 5")
```

```python
# Export your final cleaned main dataset and customer summary to CSV files for download

# 1. Save the main cleaned transactions dataset
df.to_csv('cleaned_supermarket_sales.csv', index=False)
print("Saved: cleaned_supermarket_sales.csv")

# 2. Save the customer churn summary dataset (if created)
if 'customer_summary' in locals():
    customer_summary.to_csv('customer_churn_summary.csv', index=False)
    print("Saved: customer_churn_summary.csv")

print("\nFiles are ready for download in your Colab left sidebar (Folder icon)!")
```
